```python
display(
    spark.sql(
        """
        SELECT
            dp.product_name AS `Product Name`,
            SUM(fs.product_quantity) AS `Total Product Quantity`,
            SUM(fs.recognized_revenue) AS `Product Revenue`
        FROM lh_gold.dbo.fact_sales fs
        JOIN lh_gold.dbo.dim_product dp
            ON dp.product_id = fs.product_id
        GROUP BY
            dp.product_name
        ORDER BY `Product Revenue` DESC
        LIMIT 10
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                fs.seller_id AS `Seller`,
                SUM(fs.recognized_revenue) AS `Seller Revenue`
            FROM lh_gold.dbo.fact_sales fs
            JOIN lh_gold.dbo.dim_seller ds
                ON fs.seller_id = ds.seller_id
            WHERE fs.order_status = 'Delivered'
            GROUP BY fs.seller_id
            ORDER BY `Seller Revenue` DESC
            LIMIT 10
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                fs.order_status,
                COUNT(*) AS `Order Count`,
                ROUND(
                    COUNT(*) * 100.0 /
                    SUM(COUNT(*)) OVER (),
                    2
                ) AS `Percentage Share`
            FROM lh_gold.dbo.fact_sales fs
            GROUP BY fs.order_status
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                dp.product_id,
                SUM(fs.product_quantity) AS total_product_quantity
            FROM lh_gold.dbo.dim_product dp
            LEFT JOIN lh_gold.dbo.fact_sales fs
                ON dp.product_id = fs.product_id
            GROUP BY dp.product_id
            ORDER BY total_product_quantity DESC
            LIMIT 10
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                dd.year_number,
                ROUND(
                    SUM(fs.recognized_revenue),
                    2
                ) AS revenue
            FROM lh_gold.dbo.fact_sales fs
            JOIN lh_gold.dbo.dim_date dd
                ON fs.date_key = dd.date_key
            WHERE fs.order_status = 'Delivered'
            GROUP BY dd.year_number
            ORDER BY dd.year_number
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                CONCAT(dd.month_number, '/', dd.year_number) AS month_year,
                COUNT(DISTINCT fs.order_id) AS total_orders_by_month_year
            FROM lh_gold.dbo.fact_sales fs
            JOIN lh_gold.dbo.dim_date dd
                ON fs.date_key = dd.date_key
            GROUP BY
                dd.month_number,
                dd.year_number
            ORDER BY
                dd.year_number,
                dd.month_number
        """
    )
)
```

```python
display(
    spark.sql(
        """
            SELECT
                order_status,
                COUNT(DISTINCT order_id) AS order_count,
                ROUND(
                    COUNT(DISTINCT order_id) * 100.0
                    / SUM(COUNT(DISTINCT order_id)) OVER (),
                    2
                ) AS percentage_share
            FROM lh_gold.dbo.fact_sales
            GROUP BY order_status
            ORDER BY order_count DESC
        """
    )
)
```
